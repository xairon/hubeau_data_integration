# Test Complet du Wrapper Hub'Eau

Ce notebook teste le wrapper Hub'Eau en extrayant des échantillons de données de chaque API.

**Objectifs** :
- Tester les 8 APIs Hub'Eau
- Extraire des données sur courtes périodes
- Exporter en CSV et Parquet
- Tester le pipeline DuckDB
- Créer des visualisations basiques

**Durée estimée** : 5-10 minutes

```python
# Cellule 1: Setup et Imports
import sys
sys.path.insert(0, '../src')

from hubeau import HubeauClient, HubeauPipeline
import pandas as pd
from datetime import datetime, timedelta
import os
from pathlib import Path

# Créer répertoire pour les exports
output_dir = Path("../data/test_exports")
output_dir.mkdir(parents=True, exist_ok=True)

print("Imports reussis")
print(f"Repertoire d'export: {output_dir}")
```

```python
# Cellule 2: Initialisation du Client
print("="*60)
print("INITIALISATION DU CLIENT HUB'EAU")
print("="*60)

client = HubeauClient()

print("\nAPIs disponibles:")
for i, api in enumerate(client.list_apis(), 1):
    print(f"  {i}. {api}")
```

```python
# Cellule 3: Test 1 - Piézométrie (Stations)
print("="*60)
print("TEST 1: PIEZOMETRIE - Stations")
print("="*60)

data_piezo_stations = list(client.get_data(
    api="piezometry",
    endpoint="stations",
    limit=10
))

df_piezo_stations = pd.DataFrame(data_piezo_stations)

print(f"Nombre de stations: {len(df_piezo_stations)}")
print(f"Colonnes: {list(df_piezo_stations.columns)[:5]}...")
print("\nApercu:")
display(df_piezo_stations.head())

csv_path = output_dir / "piezometry_stations.csv"
df_piezo_stations.to_csv(csv_path, index=False)
print(f"\nExporte vers: {csv_path}")
```

```python
# Cellule 4: Test 2 - Piézométrie (Chroniques - 7 derniers jours)
print("="*60)
print("TEST 2: PIEZOMETRIE - Chroniques (7 derniers jours)")
print("="*60)

date_fin = datetime.now()
date_debut = date_fin - timedelta(days=7)

if len(df_piezo_stations) > 0 and 'code_bss' in df_piezo_stations.columns:
    code_bss = df_piezo_stations.iloc[0]['code_bss']
    
    data_piezo_chroniques = list(client.get_data(
        api="piezometry",
        endpoint="chroniques",
        params={
            "code_bss": code_bss,
            "date_debut_mesure": date_debut.strftime("%Y-%m-%d"),
            "date_fin_mesure": date_fin.strftime("%Y-%m-%d")
        },
        limit=100
    ))
    
    df_piezo_chroniques = pd.DataFrame(data_piezo_chroniques)
    
    print(f"Station: {code_bss}")
    print(f"Nombre de mesures: {len(df_piezo_chroniques)}")
    print("\nApercu:")
    display(df_piezo_chroniques.head())
    
    csv_path = output_dir / "piezometry_chroniques.csv"
    df_piezo_chroniques.to_csv(csv_path, index=False)
    print(f"\nExporte vers: {csv_path}")
else:
    print("Aucune station disponible pour tester les chroniques")
```

```python
# Cellule 12: Test du Pipeline DuckDB
print("="*60)
print("TEST PIPELINE: Export vers DuckDB")
print("="*60)

pipeline = HubeauPipeline(
    destination="duckdb",
    dataset_name="test_notebook"
)

print(f"Pipeline cree")
print(f"  Nom: {pipeline.pipeline_name}")
print(f"  Destination: {pipeline.destination_type}")

print("\nChargement de donnees piezometrie...")
result = pipeline.load(
    api="piezometry",
    endpoint="stations",
    params={"size": 20}
)

print(f"Donnees chargees dans DuckDB")
info = pipeline.get_pipeline_info()
print(f"  Pipeline: {info['working_dir']}")
```

```python
# Cellule 13: Test Export Parquet
print("="*60)
print("TEST: Export Parquet")
print("="*60)

if len(df_piezo_stations) > 0:
    parquet_path = output_dir / "piezometry_stations.parquet"
    
    df_piezo_stations.to_parquet(parquet_path, index=False)
    
    print(f"Exporte vers: {parquet_path}")
    print(f"  Taille: {parquet_path.stat().st_size / 1024:.2f} KB")
    
    df_reloaded = pd.read_parquet(parquet_path)
    print(f"Verification: {len(df_reloaded)} lignes relues")
```

```python
# Cellule 14: Résumé des Exports
print("="*60)
print("RESUME DES EXPORTS")
print("="*60)

csv_files = list(output_dir.glob("*.csv"))
parquet_files = list(output_dir.glob("*.parquet"))

print(f"\nFichiers CSV exportes: {len(csv_files)}")
for f in sorted(csv_files):
    size_kb = f.stat().st_size / 1024
    print(f"  - {f.name:40s} ({size_kb:6.2f} KB)")

print(f"\nFichiers Parquet exportes: {len(parquet_files)}")
for f in sorted(parquet_files):
    size_kb = f.stat().st_size / 1024
    print(f"  - {f.name:40s} ({size_kb:6.2f} KB)")

print(f"\nTous les exports dans: {output_dir}")
```

```python
# Cellule 15: Test des Fonctions Utilitaires
print("="*60)
print("TEST: Fonctions Utilitaires")
print("="*60)

print("\n1. Endpoints disponibles pour 'piezometry':")
endpoints = client.list_endpoints("piezometry")
for ep in endpoints:
    print(f"  - {ep}")

print("\n2. Informations pipeline:")
info = pipeline.get_pipeline_info()
for key, value in info.items():
    print(f"  {key}: {value}")
```

```python
# Cellule 16: Visualisation - Carte des Stations Piézométriques
print("="*60)
print("VISUALISATION: Carte des Stations Piezometriques")
print("="*60)

import matplotlib.pyplot as plt

if 'latitude' in df_piezo_stations.columns and 'longitude' in df_piezo_stations.columns:
    plt.figure(figsize=(10, 8))
    plt.scatter(
        df_piezo_stations['longitude'], 
        df_piezo_stations['latitude'],
        alpha=0.6,
        s=100,
        c='blue',
        edgecolors='black'
    )
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')
    plt.title('Localisation des Stations Piezometriques (Echantillon)')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    map_path = output_dir / 'map_piezometry_stations.png'
    plt.savefig(map_path, dpi=150)
    plt.show()
    
    print(f"Carte sauvegardee: {map_path}")
else:
    print("Pas de coordonnees disponibles pour la visualisation")
```

```python
# Cellule 17: Conclusion
print("="*80)
print("TESTS TERMINES AVEC SUCCES")
print("="*80)

csv_files = list(output_dir.glob("*.csv"))

print("\nResume:")
print(f"  - 8 APIs testees")
print(f"  - {len(csv_files)} fichiers CSV generes")
print(f"  - Pipeline DuckDB teste")
print(f"  - Exports Parquet valides")
print(f"  - Tous les fichiers dans: {output_dir}")

print("\nLe wrapper Hub'Eau est pleinement fonctionnel!")
print("\nProchaines etapes suggerees:")
print("  1. Tester avec des periodes plus longues")
print("  2. Charger vers PostgreSQL (si disponible)")
print("  3. Implementer des transformations custom")
print("  4. Creer des dashboards de visualisation")
```
